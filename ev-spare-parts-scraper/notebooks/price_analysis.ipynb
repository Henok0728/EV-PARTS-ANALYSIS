### 5. Strategic Import Prioritization Conclusions

1. **Tier 1 (Immediate Air / Fast Ocean Import - Score $\ge 4.2$):**
   - **Cabin Air Filters & Wiper Blades:** Highest sourcing practicality ($5.0/5.0$), compact volume, zero freight penalty, frequent replacement cycle in dusty Addis Ababa conditions.
   - **Brake Pads:** Essential safety wear item with high demand ($4.8/5.0$). Compatible across shared manufacturer platforms (e.g., BYD Seagull / Dolphin / E2 and VW ID.4 / ID.6).
   - **Side Mirrors:** High breakage and theft vulnerability in crowded urban traffic. High scarcity in genuine spec with BSD (Blind Spot Detection).

2. **Tier 2 (Planned Ocean Container Import - Score $3.5 - 4.1$):**
   - **Shock Absorbers:** Critical wear item given rough road surfaces and speed bumps; high unit value (ETB 70,000 - 90,000/pair).
   - **Brake Discs / Rotors:** Heavy cast iron parts that require bulk ocean shipment to maintain economic viability.
   - **Headlights & Tail Lights:** High unit price (ETB 85,000 - 240,000) with moderate fragility; requires careful crating.

3. **Tier 3 (Bulk / On-Demand Sourcing - Score $< 3.5$):**
   - **Bumpers & Fenders:** Severe volumetric dimensional weight penalties when shipped unpainted. High scarcity locally, best imported in complete 40ft container consignments.
